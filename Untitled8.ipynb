```python
@dataclass
class WorkflowState:
    topic: str

    retrieved_chunks: List[str] = field(default_factory=list)

    extracted_points: List[str] = field(default_factory=list)

    synthesis: str = ""

    final_report: str = ""

    completed_steps: List[str] = field(default_factory=list)

    error: Optional[str] = None

    error_step: Optional[str] = None

    created_at: str = field(
        default_factory=lambda: datetime.now().isoformat()
    )
```

```python
CHECKPOINT_DIR = Path("checkpoints")
CHECKPOINT_DIR.mkdir(exist_ok=True)

def checkpoint_file(topic: str, step: str) -> Path:
    safe_topic = "".join(
        c if c.isalnum() else "_"
        for c in topic
    ).strip("_")

    return CHECKPOINT_DIR / f"{safe_topic}_{step}.json"
```

```python
def save_checkpoint(state: WorkflowState, step: str):
    path = checkpoint_file(state.topic, step)

    data = asdict(state)

    with open(path, "w", encoding="utf-8") as f:
        json.dump(
            data,
            f,
            indent=2,
            ensure_ascii=False
        )

    print(f"✅ Checkpoint saved: {path}")

    return path
```

```python
def load_checkpoint(topic: str, step: str):
    path = checkpoint_file(topic, step)

    if not path.exists():
        return None

    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    state = WorkflowState(**data)

    print(f"📂 Checkpoint loaded: {path}")

    return state
```

```python
STEP_ORDER = [
    "step1_search",
    "step2_extract",
    "step3_synthesis",
    "step4_report"
]


def get_latest_checkpoint(topic: str):

    for step in reversed(STEP_ORDER):

        path = checkpoint_file(topic, step)

        if path.exists():
            return step

    return None
```

```python
import time
from google.genai.errors import ClientError

def search_sources(topic: str, max_retries: int = 3) -> list[str]:
    print("\n🔎 STEP 1 — SEARCHING SOURCES")
    print("Topic:", topic)

    prompt = f"""
Act as a research analyst. Retrieve key research literature and reference points on:
"{topic}"

Provide 4 to 6 detailed, factual source summaries. For each source, include:
- Source / Publication Title
- Author / Institution
- Key findings, metrics, and core argument
- Practical implications

Format each source as a standalone paragraph, separated by two newlines (\n\n).
"""

    for attempt in range(max_retries):
        try:
            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0.2,
                    max_output_tokens=2048
                )
            )

            text = response.text or ""
            chunks = [chunk.strip() for chunk in text.split("\n\n") if len(chunk.strip()) > 40]

            print(f"✅ Step 1 complete: Retrieved {len(chunks)} research source chunks.")
            return chunks

        except ClientError as e:
            if "429" in str(e) or "RESOURCE_EXHAUSTED" in str(e):
                wait_time = (attempt + 1) * 20  # Wait 20s, 40s, 60s
                print(f"⚠️ Quota rate limit reached (429). Cooling down for {wait_time}s (Attempt {attempt+1}/{max_retries})...")
                time.sleep(wait_time)
            else:
                raise e

    raise RuntimeError("Failed to complete Step 1 after retries due to quota exhaustion.")
```

```python
import time
from google.genai.errors import ServerError, ClientError

# Alternate models to try in case of 503 high demand
FALLBACK_MODELS = ["gemini-3.8-flash", "gemini-2.5-pro"]

def safe_generate(contents: str, tools=None, temperature: float = 0.2) -> str:
    """Handles 503 high demand and 429 quota errors gracefully with automatic retries and model fallbacks."""
    for model_candidate in FALLBACK_MODELS:
        for attempt in range(3):
            try:
                config_args = {"temperature": temperature}
                if tools:
                    config_args["tools"] = tools

                response = client.models.generate_content(
                    model=model_candidate,
                    contents=contents,
                    config=types.GenerateContentConfig(**config_args)
                )
                if response and response.text:
                    return response.text
            except (ServerError, ClientError) as e:
                err_msg = str(e)
                if "503" in err_msg or "UNAVAILABLE" in err_msg:
                    wait = 6 * (attempt + 1)
                    print(f"⏳ Server 503 on {model_candidate}. Retrying in {wait}s...")
                    time.sleep(wait)
                elif "429" in err_msg or "RESOURCE_EXHAUSTED" in err_msg:
                    wait = 15 * (attempt + 1)
                    print(f"⏳ Quota limit 429 on {model_candidate}. Cooling down {wait}s...")
                    time.sleep(wait)
                else:
                    print(f"⚠️ Model {model_candidate} error: {err_msg[:120]}")
                    break  # Switch to next fallback model

    raise RuntimeError("All models and retry attempts failed due to Google backend overload.")
```

```python
def search_sources(topic: str) -> list[str]:
    print(f"\n🔎 STEP 1 — SEARCHING SOURCES: {topic}")
    prompt = f"""
Act as a research analyst. Retrieve key research literature and reference points on:
"{topic}"

Provide 4 to 6 detailed, factual source summaries. For each source, include:
- Source / Publication Title
- Author / Institution
- Key findings, metrics, and core argument
- Practical implications

Format each source as a standalone paragraph, separated by two newlines.
"""
    raw_text = safe_generate(prompt, temperature=0.2)
    chunks = [c.strip() for c in raw_text.split("\n\n") if len(c.strip()) > 35]
    print(f"✅ Step 1 complete: Retrieved {len(chunks)} source chunks.")
    return chunks

def extract_key_points(chunks: list[str]) -> list[str]:
    print("\n🧠 STEP 2 — EXTRACTING KEY POINTS")
    combined_sources = "\n\n".join(f"CHUNK {i+1}:\n{chunk}" for i, chunk in enumerate(chunks))

    prompt = f"""
Analyze the following research chunks and extract 6 to 10 concrete, factual bullet points.
Preserve key technical data, metrics, and findings. Avoid duplicates.
Return ONLY a numbered list.

Sources:
{combined_sources}
"""
    raw_text = safe_generate(prompt, temperature=0.2)
    points = []
    for line in raw_text.splitlines():
        line = line.strip()
        if line and line[0].isdigit():
            points.append(line.lstrip("0123456789.- "))
    print(f"✅ Step 2 complete: Extracted {len(points)} key points.")
    return points

def synthesise_findings(points: list[str]) -> str:
    print("\n🔗 STEP 3 — SYNTHESIZING FINDINGS")
    formatted_points = "\n".join(f"- {p}" for p in points)

    prompt = f"""
Synthesize the following points into a structured narrative:
{formatted_points}

Include:
1. Core Overview
2. Primary Trends & Drivers
3. Technical/Systemic Bottlenecks
4. Strategic Outlook
"""
    synthesis = safe_generate(prompt, temperature=0.3)
    print("✅ Step 3 complete: Synthesis compiled.")
    return synthesis

def format_report(topic: str, synthesis: str) -> str:
    print("\n📝 STEP 4 — FORMATTING FINAL REPORT")
    prompt = f"""
Create an executive research publication.
Topic: {topic}
Synthesis:
{synthesis}

Required Markdown sections:
# Research Report: {topic}
## Executive Summary
## Background & Context
## Technical Findings
## Strategic Bottlenecks
## Future Trajectory & Conclusion
"""
    report = safe_generate(prompt, temperature=0.2)
    print("✅ Step 4 complete: Final report formatted.")
    return report
```

```python
import json
import time
from dataclasses import asdict

def run_workflow(topic: str, simulate_failure: bool = False) -> WorkflowState:
    print("=" * 60)
    print(f"🚀 RUNNING WORKFLOW: {topic}")
    print("=" * 60)

    # 1. Check for existing checkpoints
    latest_step = find_latest_checkpoint(topic)
    if latest_step:
        state = load_checkpoint(topic, latest_step)
        print(f"🔄 Resumed state: Completed steps: {state.completed_steps}")
    else:
        state = WorkflowState(topic=topic)
        print("🌱 Starting fresh.")

    try:
        # Step 1: Search
        if "step1_search" not in state.completed_steps:
            state.retrieved_chunks = search_sources(topic)
            state.completed_steps.append("step1_search")
            save_checkpoint(state, "step1_search")
            time.sleep(2)
        else:
            print("⏭️ Step 1 already complete. Skipping.")

        # Step 2: Extract
        if "step2_extract" not in state.completed_steps:
            state.extracted_points = extract_key_points(state.retrieved_chunks)
            state.completed_steps.append("step2_extract")
            save_checkpoint(state, "step2_extract")
            time.sleep(2)
        else:
            print("⏭️ Step 2 already complete. Skipping.")

        # Step 3: Synthesize (with simulated failure trigger)
        if "step3_synthesis" not in state.completed_steps:
            if simulate_failure:
                print("\n💥 [TEST TRIGGER] Simulating runtime crash inside Step 3!")
                raise RuntimeError("Controlled crash for Day 28 recovery verification.")

            state.synthesis = synthesise_findings(state.extracted_points)
            state.completed_steps.append("step3_synthesis")
            save_checkpoint(state, "step3_synthesis")
            time.sleep(2)
        else:
            print("⏭️ Step 3 already complete. Skipping.")

        # Step 4: Final Report
        if "step4_report" not in state.completed_steps:
            state.final_report = format_report(topic, state.synthesis)
            state.completed_steps.append("step4_report")
            save_checkpoint(state, "step4_report")
        else:
            print("⏭️ Step 4 already complete. Skipping.")

        print("\n🎉 WORKFLOW SUCCEEDED!")
        return state

    except Exception as e:
        state.error = str(e)
        state.error_step = "step3_synthesis" if simulate_failure else "unknown"

        err_path = get_checkpoint_path(topic, "partial_results")
        with open(err_path, "w", encoding="utf-8") as f:
            json.dump(asdict(state), f, indent=2, ensure_ascii=False)

        print(f"\n❌ WORKFLOW HALTED: {e}")
        print(f"⚠️ Partial state saved to: {err_path.name}")
        return state
```

```python
from pathlib import Path
import json
from dataclasses import asdict

CHECKPOINT_DIR = Path("checkpoints")
CHECKPOINT_DIR.mkdir(exist_ok=True)

STEP_ORDER = [
    "step1_search",
    "step2_extract",
    "step3_synthesis",
    "step4_report"
]

def get_checkpoint_path(topic: str, step: str) -> Path:
    safe_topic = "".join(c if c.isalnum() else "_" for c in topic).strip("_")
    return CHECKPOINT_DIR / f"{safe_topic}_{step}.json"

def save_checkpoint(state: WorkflowState, step: str) -> Path:
    path = get_checkpoint_path(state.topic, step)
    data = asdict(state)
    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)
    print(f"💾 Checkpoint saved: {path.name}")
    return path

def load_checkpoint(topic: str, step: str):
    path = get_checkpoint_path(topic, step)
    if not path.exists():
        return None
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)
    print(f"📂 Checkpoint loaded from: {path.name}")
    return WorkflowState(**data)

def find_latest_checkpoint(topic: str):
    for step in reversed(STEP_ORDER):
        path = get_checkpoint_path(topic, step)
        if path.exists():
            return step
    return None

print("✅ Checkpoint helper functions registered successfully!")
```
